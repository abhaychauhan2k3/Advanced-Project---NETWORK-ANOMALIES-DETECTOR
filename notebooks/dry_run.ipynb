```python
import joblib
import numpy as np 
```

```python
import pandas as pd
df = pd.read_csv(r'E:\Anomaly Detector\data\cicids2017_cleaned.csv')
df.columns
```

```python
rf_model = joblib.load(r'E:\Anomaly Detector\models\Random_forest_model.pkl')
scaler = joblib.load(r'E:\Anomaly Detector\models\scaler.pkl')
```

```python
sample_features = np.array([[80,2586924,4,24,6,6,6.0,0.0,0,0,0.0,0.0,9.277427555,1.546237926,862308.0,1491805.311,2584896,3,2586924,862308.0,1491805.311,2584896,3,0,0.0,0.0,0,0,80,0,1.546237926,0.0,6,6,6.0,0.0,0.0,0,0,1,7.5,24,256,-1,3,20,0.0,0,0,0.0,0,0]])
sample_scaled = scaler.transform(sample_features)
```

```python
label_map = {'Bots': 0, 'Brute Force': 1, 'DDoS': 2, 'DoS': 3, 'Normal Traffic': 4, 'Port Scanning': 5, 'Web Attacks': 6}
# Reverse the label_map to get integer-to-class name mapping
reverse_label_map = {v: k for k, v in label_map.items()}

pred_labels_encoded = rf_model.predict(sample_scaled)
pred_labels_names = [reverse_label_map[int(x)] for x in pred_labels_encoded]

for i, label in enumerate(pred_labels_names):
    print(f"Sample {i+1} -> Predicted Class : {label}")
```
